##Ejercicio 1

Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
# Probabilidad de que la suma de dos dados sea exactamente 5

casos_posibles = 0
casos_favorables = []

for dado1 in range(1, 7):
    for dado2 in range(1, 7):
        casos_posibles += 1
        if dado1 + dado2 == 5:
            casos_favorables.append((dado1, dado2))

probabilidad = len(casos_favorables) / casos_posibles

print("Casos posibles:", casos_posibles)
print("Casos favorables:", len(casos_favorables))
print("Resultados favorables:", casos_favorables)
print("Probabilidad:", round(probabilidad * 100, 2), "%")
```

##Ejercicio 2

En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.

```python
# Probabilidad de que una contraseña de 3 dígitos no repita ninguno

from itertools import product

digitos = "0123456789"
casos_posibles = 0
casos_favorables = 0

for contraseña in product(digitos, repeat=3):
    casos_posibles += 1
    if len(set(contraseña)) == 3:
        casos_favorables += 1

probabilidad = casos_favorables / casos_posibles

print("Contraseñas posibles:", casos_posibles)
print("Contraseñas sin repeticiones:", casos_favorables)
print("Probabilidad:", round(probabilidad * 100, 2), "%")
```

##Ejercicio 3

Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

```python
# Probabilidad de que la favorita quede primera
# y la menos favorita quede última

from math import factorial

canciones = 8

casos_posibles = factorial(canciones)
casos_favorables = factorial(canciones - 2)

probabilidad = casos_favorables / casos_posibles

print("Órdenes posibles:", casos_posibles)
print("Órdenes favorables:", casos_favorables)
print("Probabilidad:", round(probabilidad * 100, 2), "%")
```

##Ejercicio 4

Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o que la suma sea 10? ¿Por cuántos resultados le gana una suma a la otra?

```python
# Comparar la probabilidad de obtener suma 9 y suma 10

casos_posibles = 0
suma_9 = 0
suma_10 = 0

for dado1 in range(1, 7):
    for dado2 in range(1, 7):
        for dado3 in range(1, 7):
            casos_posibles += 1
            suma = dado1 + dado2 + dado3

            if suma == 9:
                suma_9 += 1
            elif suma == 10:
                suma_10 += 1

print("Casos posibles:", casos_posibles)
print("Resultados con suma 9:", suma_9)
print("Resultados con suma 10:", suma_10)
print("Probabilidad de suma 9:", round(suma_9 / casos_posibles * 100, 2), "%")
print("Probabilidad de suma 10:", round(suma_10 / casos_posibles * 100, 2), "%")
print("Diferencia de resultados:", abs(suma_10 - suma_9))

if suma_9 > suma_10:
    print("La suma de 9 es más probable.")
elif suma_10 > suma_9:
    print("La suma de 10 es más probable.")
else:
    print("Ambas sumas son igual de probables.")
```

##Ejercicio 5
Ana, Bruno, Carla y Diego hacen un amigo secreto: cada quien saca un regalo al azar y todos los repartos son igualmente probables. Calcula la probabilidad de que nadie saque su propio regalo.

```python
# Probabilidad de que nadie saque su propio regalo

from itertools import permutations

personas = ["Ana", "Bruno", "Carla", "Diego"]
repartos = permutations(personas)

casos_posibles = 0
casos_favorables = 0

for reparto in repartos:
    casos_posibles += 1
    nadie_recibe_su_regalo = True

    for i in range(len(personas)):
        if personas[i] == reparto[i]:
            nadie_recibe_su_regalo = False
            break

    if nadie_recibe_su_regalo:
        casos_favorables += 1

probabilidad = casos_favorables / casos_posibles

print("Personas:", personas)
print("Repartos posibles:", casos_posibles)
print("Repartos favorables:", casos_favorables)
print("Probabilidad:", f"{casos_favorables}/{casos_posibles}")
print("Porcentaje:", round(probabilidad * 100, 2), "%")
```
